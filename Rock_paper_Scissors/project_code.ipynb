```python
# declare a variable  , List of valid choices
# player to make a choice Rock , paper, or scissors? (r/p/s):
# if choice is not valid
#   print error
# print user choice
# Let the  computer to make a choice (rock , paper, scissors)
# print computer choice (emojis)
# if you have rock & computer have scissors and you have scissors and computer have paper and you have paper and computer have rock (human win)
# else computer win 

# Dictionary data structure we will map a key with a value
# Key value pair

# key -> value
# 'r' -> '🪨'

# Refactoring : Changing the structure of the code without changing  its functionality
# Common refactoring method is Modularization 
# Modularization: Breaking down a large program into smaller reusable parts called modules or functions

```

```python
import random 

# key -> value
# 'r' -> '🪨'

emojis = {'r':'🪨', 'p':'📃', 's':'✂️'}
# tuple of valid choices -> Because tuples are immutable
choices = ('r','p','s') 

def get_user_choice():
    while True:
        user_choice = input('Rock, paper, or scissors? (r/p/s): ').lower() # here we are using the lower method
        # so we can treat user input as lower case text
        if user_choice not in choices:
            print("Invalid choice!")
            continue
        else:
             break
    return user_choice

def display_choices(user_choice, computer_choice):
    # print('You chose: '+user_choice)
    print(f'You chose: {emojis[user_choice]} ') # if we will use formatted string
    print(f'Computer choice is: {emojis[computer_choice]} ')
     

def determine_winner(user_choice, computer_choice):
    if user_choice == computer_choice:
        print('Its a Tie')
    elif (user_choice == 'r' and computer_choice == 's') or (user_choice == 's' and computer_choice == 'p') or (user_choice == 'p' and computer_choice == 'r') :
        print ("User is winner 🎉")
    else:
        print("Computer is winner 🥲")           

def play_game():
    while True:
        user_choice = get_user_choice()
        # here we will use choice method so that we can pass a list inside it and we can choose particular choices from the elements
        computer_choice = random.choice(choices)
        

        display_choices(user_choice, computer_choice)

        determine_winner(user_choice,computer_choice)
     
        should_continue = input('Continue? (y/n): ').lower()
        if should_continue == 'n':
            break    

play_game()
```
