```python
import random

playing = True
while playing:  # until playing is True
    ques =input( "Roll the dice ?")
    if (ques == 'y' or ques == 'Y'):
    
        # generate  a random pair of dice
        dice1 = random.randint(1,6)  # randint is used to generate a random integer
        dice2 = random.randint(1,6)
        print(f"{dice1}, {dice2}")
    elif(ques == 'n' or ques == 'N'):
        print("Thanks for playing ")
        break
    else:
        print("Invalid choice")
```

```python
import random

playing = True
while playing:  # until playing is True
    ques =input( "Roll the dice ?")
    count = 0;
    if (ques == 'y' or ques == 'Y'):
        dice_count = int(input('How many dice you want to roll out: '))
        while dice_count > 0:
            # generate  a random pair of dice
            dice1 = random.randint(1,6)  # randint is used to generate a random integer
            dice2 = random.randint(1,6)
            print(f"{dice1}, {dice2}")
            dice_count -= 1
            count += 1
        print("Number of times dice rolled",count)    
           
    elif(ques == 'n' or ques == 'N'):
        print("Thanks for playing ")
        break
    else:
        print("Invalid choice")
```
