```python

# Generate a random number
import random

number_to_guess = random.randint(1, 100)
# loop
while True:
    try:
    # if we entered a invalid value it will throw and error
    # so to handle this error we will write this portion of code inside a try block
        guess =int( input("Guess the number between 1 and 100: "))
        if  guess < number_to_guess :
            print("The number is Too low:",guess)
                
        elif guess > number_to_guess :
            print ("The number is Too high: ",guess)         
        else: 
            print("the number is valid , well done:", guess) 
            break
    except ValueError:    
        print ("please enter a valid number")

       
```
